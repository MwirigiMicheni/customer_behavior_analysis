## Clean Data

```python
df.isnull().sum()
```

Notice the 37 null values in the "Review Rating" column. These could be filled by the mean or the median. Preferably median.

Nonetheless, automatic filling using median should be carefully considered given that the products without reviews could be of varying categories.

Therefore, for current dataset, filling will be done by imputing the median review rating from the respective categories.

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

Check if values have been filled

```python
df.isnull().sum()
```

Revise the column names to ease working with data. Also remove the trailing "(usd)" in the "Purchase Amount (USD)" column.

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
df.columns
```

## Feature Engineering

Create a new column named "age_group"

```python
age_group_labels = ['Young Adult', 'Adult', 'Middle Aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = age_group_labels)
df[['age', 'age_group']].head(10)
```

```python
df['frequency_of_purchases'].unique()
```

```python
frequency_map = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Annually': 365,
    'Quarterly': 90,
    'Bi-Weekly': 3,
    'Monthly': 30,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] =df['frequency_of_purchases'].map(frequency_map)
df[['frequency_of_purchases', 'purchase_frequency_days']].head(10)
```

Check for redundancy between "discount_applied" and "promo_code_used" columns. Logically, promo codes are sort of discounts and clients often get discounts when using promo codes. Therefore, there could be similarities between the two columns.

```python
df[['discount_applied', 'promo_code_used']].head(10)
```

Overview of first ten rows shows similarity across both columns. Check for all.

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

"promo_code_used" will be dropped.

```python
df = df.drop('promo_code_used', axis=1)
df.columns
```
